### PDF

```python
from langchain_community.document_loaders import PyPDFLoader

load=PyPDFLoader('./data/file.pdf')
```

```python
pdf_docs=load.load()
pdf_docs
```

### Web

```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
load=WebBaseLoader(web_paths=('https://www.geeksforgeeks.org/deep-learning/deep-learning-tutorial/',),
                   bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                       class_=('ArticleHeader_article-title__futDC')
                   ))
                   )
```

```python
docs=load.load()
docs
```
