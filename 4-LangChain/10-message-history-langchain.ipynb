```python
import os
from dotenv import load_dotenv
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

model=ChatGroq(model="llama-3.1-8b-instant",groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage

model.invoke([HumanMessage(content="Hi, my name is Mouryagna and I am from IIIT Nagpur")])
```

```python
model.invoke([
    HumanMessage(content="Hi, my name is Mouryagna and I am from IIIT Nagpur"),
    AIMessage(content="Nice to meet you, Mouryagna! IIIT Nagpur is a well-known institution in the field of Information Technology and related disciplines. What brings you here today? Do you have any questions or topics you'd like to discuss? I'm all ears and happy to help."),
    HumanMessage(content="Hey what's my name and where i am from ?")
])
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
store={}
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

with_message_history=RunnableWithMessageHistory(model,get_session_history)
config={"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [HumanMessage(content="Hi, my name is Mouryagna and I am from IIIT Nagpur")],config=config
)
response
```

```python
with_message_history.invoke(
    [HumanMessage(content="Hey what's my name and where i am from ?")],config=config
)
```

```python
#chaging config
config1={"configurable":{"session_id":"chat1"}}
with_message_history.invoke(
    [HumanMessage(content="Hey what's my name and where i am from ?")],config=config1
)
```

```python
#chaging session
config1={"configurable":{"session_id":"chat2"}}
with_message_history.invoke(
    [HumanMessage(content="Hey what's my name and where i am from ?")],config=config1
)
```

### Prompt template

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
```

```python
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpful assistant. Answer all the following question to the nest to you ability"),
        MessagesPlaceholder(variable_name="messages")
    ]
)
```

```python
chain= prompt | model
```

```python
chain.invoke({"messages":[HumanMessage(content="Hi, my name is Mouryagna and I am from IIIT Nagpur")]})
```

```python
with_message_history=RunnableWithMessageHistory(chain,get_session_history)
```

```python
config={"configurable":{"session_id":"chat1"}}
response=with_message_history.invoke(
    [HumanMessage(content="Hey what's my name and where i am from ?")],config=config
)
response
```

```python
with_message_history=RunnableWithMessageHistory(chain,get_session_history,input_messages_key="messages")
config={"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke({
    "messages":[HumanMessage(content="Hey, I am mouryagna and I am from IIIT Nagpur")]
},config=config)
```
