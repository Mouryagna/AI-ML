```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')
os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGCHAIN_API_KEY')
os.environ['LANGCHAIN_TRACING_V2']="true"
os.environ['LANGCHAIN_PROJECT']='LANGCHAIN_PROJECT'
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings


loader=WebBaseLoader("https://docs.langchain.com/langsmith/prompt-engineering-quickstart")
docs=loader.load()

text_splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
docs=text_splitter.split_documents(docs)

embeddings=OpenAIEmbeddings()

db=FAISS.from_documents(docs,embeddings)
```

```python
query="What is Prompt Engineering?"

response=db.similarity_search(query)
response
```

```python
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

prompt=ChatPromptTemplate.from_template(
    """
    Answer the following question based on the provided context:
    <context> {context} <context>
    """
)
llm=ChatOpenAI(model="gpt-4o")
document_chain=create_stuff_documents_chain(llm,prompt)
```

```python
from langchain_core.documents import Document

document_chain.invoke({
    "input":"What is Langsmith Prompt Engineering?",
    "context": [Document(page_content="About Langsmith Prompt Engineering")]
})
```

```python
from langchain.chains import create_retrieval_chain
```

```python
retriever=db.as_retriever()
retriever_chain=create_retrieval_chain(retriever,document_chain)
```

```python
response=retriever_chain.invoke({
    "input":"What is Langsmith Prompt Engineering?",
})
response['context']
```
