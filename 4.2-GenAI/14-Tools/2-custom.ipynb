## Structured Tool - Pydantic

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class MultiplyInput(BaseModel):
    a: int =Field(required=True,description="First Number")
    b: int =Field(required=True,description="Second Number")
```

```python
def mul_fun(a: int,b: int)-> int:
    return a*b
```

```python
mult_tool=StructuredTool.from_function(
    func=mul_fun,
    name="multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
mult_tool.invoke({"a":3,"b":5})
```

## Basetool class

```python
from langchain_community.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

## ToolKit

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
