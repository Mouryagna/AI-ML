```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
@tool
def multiply(a:int,b:int)->int:
    """Given 2 numbers a and b this tool return their product"""
    return a*b
```

```python
multiply.invoke({"a":5,"b":5})
```

```python
llm=ChatOllama(model="llama3.1")
```

```python
llm_with_tools=llm.bind_tools([multiply])
```

```python
llm_with_tools.invoke("hi")
```

```python
query=HumanMessage("what's 8 multiply to 7")
```

```python
messages=[query]
```

```python
messages
```

```python
results=llm_with_tools.invoke(messages)
```

```python
messages.append(results)
```

```python
messages
```

```python
tool_result=multiply.invoke(results.tool_calls[0])
```

```python
messages.append(tool_result)
```

```python
messages
```

```python
llm_with_tools.invoke(messages)
```
