```python
img=cv2.imread('c.jpg')
# img[:,:,2]=0
cv2.imshow("window",img)
cv2.waitKey(0)
```

```python
img_resize=cv2.resize(img,(512,512)) #(img.shape[10]//2,img.shape[10]//2)
print(img_resize.shape)
cv2.imshow("window",img_resize)
cv2.waitKey(0)
```
