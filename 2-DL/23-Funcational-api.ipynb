```python
import tensorflow as tf
from tensorflow import keras
from keras.models import Model
```

```python
from keras.layers import *
```

```python
x=Input(shape=(3,))

hiddenLayer1=Dense(128,activation='relu')(x)
hiddenLayer2=Dense(64,activation='relu')(hiddenLayer1)

output1=Dense(1,activation='linear')(hiddenLayer2)
output2=Dense(1,activation='sigmoid')(hiddenLayer2)
```

```python
model=Model(inputs=x,outputs=[output1,output2])
```

```python
model.summary()
```

```python
from keras.utils import plot_model #visualise the model
plot_model(model)
```

```python
plot_model(model,show_shapes=True)
```
