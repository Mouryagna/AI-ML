```python
df.drop(columns=['id'], inplace=True)
```

```python
skewed_cols = df.skew(numeric_only=True)[abs(df.skew(numeric_only=True)) > 1].index

for col in skewed_cols:
    if df[col].min() > 0:
        df[col] = np.log1p(df[col])
```

```python
from sklearn.preprocessing import LabelEncoder

encode=LabelEncoder()

df['diagnosis'] = encode.fit_transform(df['diagnosis'])
```

```python
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV

x=df.drop(columns=['diagnosis'])
y=df['diagnosis']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scale=StandardScaler()

x_train=scale.fit_transform(x_train)
x_test=scale.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,f1_score,precision_score,classification_report,ConfusionMatrixDisplay,recall_score
```

```python
model=KNeighborsClassifier()
```

```python
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
print("accuracy:",accuracy_score(y_test,y_pred)*100)
print("precision:",precision_score(y_test,y_pred)*100)
print("f1:",f1_score(y_test,y_pred)*100)
print("recall:",recall_score(y_test,y_pred)*100)
```

```python
print(classification_report(y_test,y_pred))
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
```

```python
error_rates=[]
for k in range(1, 31):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train, y_train)
    y_pred = knn.predict(x_test)
    error_rates.append(np.mean(y_pred != y_test))

plt.plot(range(1, 31), error_rates, marker='o')
plt.xlabel('K Value')
plt.ylabel('Error Rate')
plt.title('Elbow Curve - Best K for KNN')
plt.show()
```

```python
knn = KNeighborsClassifier(n_neighbors=9)
knn.fit(x_train, y_train)
y_pred = knn.predict(x_test)
```

```python
print("accuracy:",accuracy_score(y_test,y_pred)*100)
print("precision:",precision_score(y_test,y_pred)*100)
print("f1:",f1_score(y_test,y_pred)*100)
print("recall:",recall_score(y_test,y_pred)*100)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
```

```python
param_grid = {
    'n_neighbors': range(1, 31),
    'metric': ['euclidean', 'manhattan', 'minkowski'],
    'weights': ['uniform', 'distance']
}

grid = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, scoring='accuracy')
grid.fit(x_train, y_train)

print("Best Params:", grid.best_params_)
print("Best Score:", grid.best_score_)
```

```python
y_pred = grid.best_estimator_.predict(x_test)
print(classification_report(y_pred, y_test))
```

```python
print("accuracy:",accuracy_score(y_test,y_pred)*100)
print("precision:",precision_score(y_test,y_pred)*100)
print("f1:",f1_score(y_test,y_pred)*100)
print("recall:",recall_score(y_test,y_pred)*100)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
```

```python
knn = KNeighborsClassifier(n_neighbors=9)

scores = cross_val_score(knn, x_train, y_train, cv=5, scoring='accuracy')

print("CV Scores:", scores)
print("Mean Accuracy:", np.mean(scores))
print("Std:", np.std(scores))
```
