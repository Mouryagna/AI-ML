```python
import numpy as np
words = ["machine", "learning", "deep", "neural", "data", "science"]

np.random.seed(42)
glove = {word: np.random.rand(50) for word in words}
```

Cosine Similarity

```python
from numpy.linalg import norm

def cosine_similarity(a, b):
    return np.dot(a, b) / (norm(a) * norm(b))
```

```python
print("Similarity (machine, learning):",
      cosine_similarity(glove["machine"], glove["learning"]))
```

```python
def most_similar(word, embeddings):
    similarities = {}

    for w in embeddings:
        similarities[w] = cosine_similarity(embeddings[word], embeddings[w])

    sorted_words = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
    return sorted_words[1:4]

print(most_similar("machine", glove))
```
