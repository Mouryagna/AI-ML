```python
from gensim.models import Word2Vec
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
```

```python
nltk.download('punkt_tab')
```

```python
text = [
    "machine learning is fun",
    "deep learning uses neural networks",
    "natural language processing is interesting",
    "word embeddings capture meaning",
    "gensim is used for word2vec",
    "python is popular for machine learning",
    "data science uses machine learning techniques"
]
```

Tokenization

```python
data = [word_tokenize(sentence.lower()) for sentence in text]
print(data)
```

```python
model = Word2Vec(
    sentences=data,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1
)
```

```python
print(model.wv.most_similar("learning"))
```

```python
print(model.wv.similarity("machine", "learning"))
```

```python
print(model.wv["machine"])
```

```python
model = Word2Vec(
    sentences=data,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0
)
```

```python
print(model.wv.most_similar("learning"))
```

```python
print(model.wv.similarity("machine", "learning"))
```
