```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'revenue':[10,100,100,0,5000,10000,2,11,55]})
```

```python
max = max(df['revenue'])
min = min(df['revenue'])
```

```python
df['revenue'].value_counts()
```

```python
temp_df = df['revenue']-min
min_scaled_df = temp_df/(max-min)
```

```python
min_scaled_df
```

```python
log_scaled_df = np.log(df['revenue']+1)
```

```python
log_scaled_df
```

```python
z_scaled_df = (df['revenue']-df['revenue'].mean())/df['revenue'].std()
```

```python
z_scaled_df
```

```python
min_scaled_df
```

```python
import matplotlib.pyplot as plt
plt.plot(df['revenue'], label='original',color='blue',marker='o')
plt.plot(min_scaled_df*100, label='min-max scaled',color='orange',marker='o')
plt.plot(log_scaled_df, label='log scaled',color='green',marker='o')
plt.plot(z_scaled_df, label='z-score scaled',color='red',marker='o')
plt.ylim(0,150)
```

```python
import numpy as np
import matplotlib.pyplot as plt

departments = ['CSE', 'ECE']

boys = [40, 30]
girls = [35, 45]

x = np.arange(len(departments))

width = 0.35

plt.bar(x - width/2, boys, width, label='Boys')
plt.bar(x + width/2, girls, width, label='Girls')

plt.xticks(x, departments)

plt.legend()

plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["CSE", "CSE", "ECE", "ECE"],
    "Marks": [80, 90, 70, 60]
})

print(df)
```

```python
grouped = df.groupby("Department").mean()

print(grouped)
```

```python
print(df.groupby("Department").count())
```

```python
df = pd.DataFrame({
    "Department": ["CSE", "CSE", "ECE", "ECE"],
    "Gender": ["M", "F", "M", "F"],
    "Marks": [80, 90, 70, 60]
})
```

```python
print(df.groupby(["Department", "Gender"]).mean())
```

```python
print(df.groupby("Department")["Marks"].mean())
```

```python
print(
    df.groupby("Department")["Marks"]
      .agg(Average="mean", Highest="max")
)
```
