```python
#scatter plot
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 40, 50, 60, 65, 75]

plt.scatter(hours,marks)
plt.xlabel("Std Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
plt.plot(hours,marks)
plt.xlabel("Std Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks Trend")
plt.show()
```

```python
departments = ['CSE', 'ECE', 'ME', 'CE']
students = [120, 90, 60, 40]

plt.bar(departments, students)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Students per Department")
plt.show()
```

```python
departments = ['CSE', 'ECE', 'ME', 'CE']
students = [120, 90, 60, 40]

plt.barh(departments, students)
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Students per Department")
plt.show()
```

```python
subjects = ['Maths', 'Physics', 'Chemistry', 'CS']
weights = [25, 20, 15, 40]

plt.pie(weights, labels=subjects, autopct='%1.1f%%')
plt.title("Subject Weightage")
plt.show()
```

```python
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 33, 35]

plt.plot(days, temperature)
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Daily Temperature Trend")
plt.show()
```

```python
days = [1, 2, 3, 4, 5]
sales = [100, 150, 220, 300, 380]

plt.fill_between(days, sales)
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Cumulative Sales")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

categories = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
values = [20, 30, 25, 35, 40]

# calculate percentages
total = sum(values)
percentages = [(v / total) * 100 for v in values]

angles = np.linspace(0, 2*np.pi, len(values), endpoint=False)

plt.figure()
ax = plt.subplot(polar=True)

bars = ax.bar(categories, values)  # your original line

# add percentage labels
for bar, pct in zip(bars, percentages):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 2,
            f'{pct:.1f}%',
            ha='center')

plt.title("Coxcomb Plot Example")
plt.show()
```
