1) To preprocess text by tokenizing, removing stopwords, and finding word frequencies.

text = "Natural Language Processing with NLTK is very interesting and powerful for text analysis."
Output: Bar chart of most frequent words.

```python
from collections import Counter
import matplotlib.pyplot as plt
import re

text = "Natural Language Processing with NLTK is very interesting and powerful for text analysis."
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

stopwords = {
    "is","and","the","a","an","with","for","to","of",
    "in","on","at","by","it","this","that","very"
}

filtered_words = [w for w in words if w not in stopwords]
freq = Counter(filtered_words)
labels, values = zip(*freq.most_common(10))

plt.figure(figsize=(8,5))
plt.bar(labels, values)
plt.xticks(rotation=45)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Most Frequent Words")
plt.tight_layout()
plt.show()
```

2) To identify the grammatical parts of speech (noun, verb, adjective, etc.) in a sentence
sentence = "The intelligent student quickly solved the complex problem using Python."
OUTPUT:
The → DT
intelligent → JJ
student → NN
quickly → RB
solved → VBD
the → DT
complex → JJ
problem → NN
using → VBG
Python → NNP


```python
import nltk
from nltk import word_tokenize, pos_tag

sentence = "The intelligent student quickly solved the complex problem using Python."
tokens = word_tokenize(sentence)
tags = pos_tag(tokens)

for word, tag in tags:
    print(word, "→", tag)
```

3) To detect named entities such as names of people, organizations, and locations.
text = "Elon Musk founded SpaceX and Tesla in the United States."
Output: Tree structure showing entities like PERSON, ORGANIZATION

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

text = "Elon Musk founded SpaceX and Tesla in the United States."
tokens = word_tokenize(text)
tagged = pos_tag(tokens)
entities = ne_chunk(tagged)
print(entities)
```
