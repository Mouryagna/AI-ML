```python
import sqlite3
```

```python
#connect to an AQLite DB

connection=sqlite3.connect('example.db')
connection
```

```python
# to Move the cursor

cursor=connection.cursor()
```

```python
#create a table
cursor.execute('''
Create Table if Not exists Employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')
# commit the Changes

connection.commit()
```

```python
cursor.execute('''
Select * from Employees
''')
```

```python
#Insert in DB
cursor.execute('''
Insert Into Employees(name,age,department)
               values('Mouryagna','20','CSA')
''')

cursor.execute('''
Insert Into Employees(name,age,department)
               values('Bob','19','CSA')
''')

cursor.execute('''
Insert Into Employees(name,age,department)
               values('Alice','18','CSE')
''')

connection.commit()
```

```python
# Query Data from DB

cursor.execute('Select * from Employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
#Update
cursor.execute('''
UPDATE Employees
Set age=21
where name='Mouryagna'
''')

connection.commit()
```

```python
cursor.execute('Select * from Employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Delete

cursor.execute('''
Delete from Employees
               where name ='Bob'
''')

cursor.execute('Select * from Employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

##### Sales Data

```python
connection=sqlite3.connect('sales_data.db')
cursor=connection.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS sales(
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT               
)
''')
```

```python
#Insert

sales_data=[
    ('2023-01-01','product1',100,'North'),
    ('2023-02-01','product2',200,'South'),
    ('2023-03-01','product3',300,'East'),
    ('2023-04-01','product4',400,'West'),
    ('2023-05-01','product5',500,'North'),
]

cursor.executemany('''
Insert into sales(date,product,sales,region)
                   values(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
cursor.execute('Select * from sales')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
#close connection
connection.close()
```

```python
def linear_search(arr, target):
    size=len(arr)
    for i in range(0,size):
        if target == arr[i]:
            return target
    return -1

lst=[1,23,4,5,8,6,3,7]
print(linear_search(lst,8))
```
